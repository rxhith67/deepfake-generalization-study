# Final figures
Generate the standardized eight-case Grad-CAM panel for each trained model.

```python
import subprocess, sys
from pathlib import Path
for model in ['meso', 'xception', 'freq_cnn', 'hybrid']:
    checkpoint = Path('../outputs/checkpoints') / f'{model}_best.pt'
    if checkpoint.exists():
        subprocess.run([sys.executable, '-m', 'src.interpret.gradcam', '--checkpoint', f'outputs/checkpoints/{model}_best.pt', '--csv', 'data/splits/test.csv', '--data-root', 'data', '--output', f'outputs/figures/{model}_gradcam.png'], cwd='..', check=True)
```
